## Data Scraping

Premier league data after VAR introduced i.e past 2018 season.

```python
premier_league = sd.Understat(leagues="ENG-Premier League", 
                              seasons=["2019/2020", "2020/2021", "2021/2022", "2022/2023", "2023/2024", "2024/2025", "2025/2026"])
```

```python
shots = premier_league.read_shot_events()
shots.head()
```

As we got the overall dataset of shots attempted in each match of premier league from season 2019 til 2025 which is the `VAR era` from `Understat` api using `soccerdata` package/sdk.

```python
shots.info()
```

```python
goals = shots[shots["result"] == 'Goal']
goals.info()
```

So total of 68007 shots only 7343 resulted in a goal.
